```python
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
import gc
```

```python
csv_path = "../cleaned_csv/merged_dataset.csv"
```

```python
# Process 500,000 rows at a time to keep RAM footprint microscopic
chunksize = 500_000 
writer = None
parquet_path = 'dataset.parquet'

print("Starting safe conversion...")

# Stream the CSV in chunks
for i, chunk in enumerate(pd.read_csv(csv_path, chunksize=chunksize)):
    print(f"Processing chunk {i + 1}...")
    
    # Convert the Pandas dataframe chunk into a PyArrow table
    table = pa.Table.from_pandas(chunk)
    
    # Initialize the Parquet writer on the very first chunk using its schema
    if writer is None:
        writer = pq.ParquetWriter(parquet_path, table.schema, compression='brotli')
    
    # Append the table chunk to the Parquet file
    writer.write_table(table)
    
    # Force memory clearance after every chunk
    del chunk
    del table
    gc.collect()

# Safely close the file connection
if writer:
    writer.close()

print("Conversion complete! Your Mac is safe.")
```
